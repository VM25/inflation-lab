# Data Audit

Verifies raw and cleaned market data: units, coverage, missing values, and the latest snapshot consistency (nominal ≈ real + breakeven).

```python
import pandas as pd

daily = pd.read_csv("../../data/processed/_clean_daily_panel.csv", parse_dates=["date"]).set_index("date")
daily.describe().T
```

```python
# units check: all rates must be decimals (< 0.20)
assert daily[["DGS2","DGS5","DGS10","DGS30"]].max().max() < 0.20
# identity check: nominal 10Y ~ real 10Y + breakeven 10Y (within 25bp)
resid = (daily["DGS10"] - daily["DFII10"] - daily["T10YIE"]).abs()
print("max |nominal - real - breakeven| (10Y):", f"{resid.max():.4%}")
daily[["DGS10","DFII10","T10YIE"]].plot(figsize=(10,4), title="10Y nominal vs real vs breakeven")
```
