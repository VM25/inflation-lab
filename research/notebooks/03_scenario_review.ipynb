# Scenario Review

Reviews scenario results: loss ordering by duration, approximation error behavior, key-rate decomposition under non-parallel shocks, and attribution reconciliation.

```python
import json, pandas as pd

scen = pd.DataFrame(json.load(open("../../data/processed/scenario_results.json")))
pivot = scen.pivot(index="scenario_id", columns="portfolio_id", values="pnl_exact")
pivot.round(0)
```

```python
# attribution must reconcile to exact P&L
err = scen.apply(lambda r: abs(sum(r["attribution"].values()) - r["pnl_exact"]), axis=1)
print("max attribution error ($):", err.max())
# convexity-adjusted approximation should beat duration-only for large parallel shocks
big = scen[scen.scenario_id == "parallel_100"]
(big.approximation_error_convexity.abs() < big.approximation_error_duration.abs()).all()
```
