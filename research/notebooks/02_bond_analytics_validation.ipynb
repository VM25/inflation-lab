# Bond Analytics Validation

Checks duration/convexity ordering, DV01 consistency (DV01 ≈ D_mod × P × 1e-4), and the finite-difference cross-validation exported by the pipeline (QuantLib-free correctness check).

```python
import pandas as pd

bonds = pd.read_csv("../../data/processed/bond_analytics.csv")
bonds
```

```python
nominal = bonds[bonds.type == "nominal"].sort_values("maturity_years")
assert nominal.duration_modified.is_monotonic_increasing
assert nominal.convexity.is_monotonic_increasing
check = (bonds.duration_modified * bonds.base_price * 1e-4 - bonds.dv01_per_100_face).abs()
print("max |DV01 - D*P*1e-4|:", check.max())
```

```python
fd = pd.read_csv("../../data/processed/_validation_finite_diff.csv")
assert (fd[["dmod_rel_diff","cvx_rel_diff"]] < 0.005).all().all()
fd
```
